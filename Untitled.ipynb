```python
import numpy as np
import matplotlib.pyplot as plt
import pylab
```

```python
x = np.linspace(0,5000000,1000)
#xB = np.linspace(0,5000000,1000)
n_regions = 13
```

```python
cost_B = 0.2 * x + 50000*n_regions
cost_A = 0.5 * x 
```

```python
pylab.plot(x,cost_A, label = "A")
pylab.plot(x,cost_B, label = "B")
pylab.plot(x, cost_B - cost_A, label = "B - A")
#pylab.plot(x)
plt.legend()
plt.show()
```
